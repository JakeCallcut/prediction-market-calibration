# Exploratory Data Analysis

### Dataset Loading and Preview

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import numpy as np

MODELLING_TABLE_PATH = Path("../data/processed/modelling_table.csv")
POLYMARKET_TABLE_PATH = Path("../data/processed/polymarket_table.csv")
KALSHI_TABLE_PATH = Path("../data/processed/kalshi_table.csv")

polymarket_df = pd.read_csv(POLYMARKET_TABLE_PATH)
kalshi_df = pd.read_csv(KALSHI_TABLE_PATH)
df = pd.read_csv(MODELLING_TABLE_PATH)
df.describe()
```

## Overall Distribution of Outcomes

```python
outcome_counts = df['outcome'].value_counts().sort_index()
plt.figure(figsize=(8, 5))
outcome_counts.plot(kind='bar', color=['#e74c3c', '#2ecc71'])
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.title('Distribution of Outcomes')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Distribution of Outcomes by Platform

```python
print(f'Polymarket predictions: {len(polymarket_df)}')
print(f'Kalshi predictions: {len(kalshi_df)}')
print(f'\nPolymarket outcome distribution:')
print(polymarket_df['outcome'].value_counts().sort_index())
print(f'\nKalshi outcome distribution:')
print(kalshi_df['outcome'].value_counts().sort_index())
```
